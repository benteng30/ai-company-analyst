```python
from google.colab import userdata
from google import genai

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

fact="Nike reported revenues of $51.4 billion in fiscal 2023. Nike is headquartered in Beaverton, Oregon."
question="What was Nike's revenue, and where are the headquarters?"
prompt = f"""answer only using the context, and if the answer isn't there, say "I don't know."
fact:{fact} question:{question}"""
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents = prompt
)
print(response.text)
```
