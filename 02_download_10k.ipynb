```python
import os
from google.colab import userdata
from edgar import Company, set_identity

set_identity(userdata.get("SEC_EMAIL"))

company = Company("NKE")
filing = company.get_filings(form="10-K").latest()
print(filing)

text = filing.markdown()
print(len(text), "characters")
print(text[:500])

os.makedirs("data", exist_ok=True)
with open("data/NKE_10K.md", "w") as f:
  f.write(text)
```

```python
filing = Company("NKE").get_filings(form = "10-K").latest()
tenk = filing.obj()

risks = tenk.risk_factors
print(len(risks), "characters")
print(risks[:1000])
print("supply chain" in risks)
print("supply chain" in risks.lower())
print(risks.lower().count("supply"))

i = risks.lower().find("supply")
print(repr(risks[i-50 : i+80]))
```
